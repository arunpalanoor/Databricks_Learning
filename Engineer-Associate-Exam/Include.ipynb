# Understanding Delta Tables

```sql
USE CATALOG fundamentals
```

Create a sample Delta table

```sql
CREATE TABLE employees (
  id INT,
  name STRING,
  salary FLOAT,
  dept STRING
)
```

```sql
INSERT INTO employees VALUES
  (1, 'Bill', 100000.0, 'Engineering'),
  (2, 'Charlie', 120000.0, 'Engineering'),
  (3, 'David', 150000.0, 'Engineering'),
  (4, 'Eve', 80000.0, 'HR'),
  (5, 'Frank', 90000.0, 'HR'),
  (6, 'Grace', 110000.0, 'HR')
```

```sql
INSERT INTO employees VALUES
  (7, 'Hannah', 130000.0, 'Engineering')
```

```sql
SELECT * FROM employees
```

```sql
DESCRIBE DETAIL employees
```

```sql
UPDATE employees
SET salary = salary+100
WHERE dept LIKE "Eng%";
SELECT * FROM employees
```
